```python

# ============================================================
# CELL 1: IMPORTS
# Musical Instrument Frequency Analyzer (B4)
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# SymPy is useful for displaying the mathematical matrix
# symbolically, although NumPy will perform the numerical
# eigenvalue calculation.
import sympy as sp

# Optional: cleaner table-style output in Colab
from IPython.display import display, Markdown

np.set_printoptions(precision=5, suppress=True)

print("Libraries imported successfully.")
```

```python

# ============================================================
# CELL 2: USER CONFIGURATION
# ============================================================

# Number of point masses used to approximate the string.
# Increasing N gives a finer approximation.
N = 20

# Physical parameters
# m = mass represented by each discrete point mass (kg)
# k = spring stiffness (N/m)
m = 0.001       # kg
k = 1000.0      # N/m

# Number of harmonics/modes to display
NUM_MODES_TO_SHOW = 3

# Length of the modeled string (m)
L = 0.65

# ------------------------------------------------------------
# OPTIONAL: If you want to model a real guitar string,
# you can change m, k and L above.
# ------------------------------------------------------------

if N < 3:
    raise ValueError("N must be at least 3.")

if m <= 0 or k <= 0 or L <= 0:
    raise ValueError("m, k and L must all be positive.")

if NUM_MODES_TO_SHOW > N:
    NUM_MODES_TO_SHOW = N

print("Project configuration")
print("---------------------")
print(f"Number of masses N       = {N}")
print(f"Mass of each point m     = {m} kg")
print(f"Spring stiffness k       = {k} N/m")
print(f"String length L          = {L} m")
print(f"Modes displayed          = {NUM_MODES_TO_SHOW}")
```
